### 6.1 Listas: `list`
Java necesita `ArrayList<Double> saldos = new ArrayList<>();`. En Python basta con corchetes, y una lista puede mezclar tipos.

````python
saldos = [100.0, 250.5, 80.0]

saldos.append(500.0)        # agregar al final  (Java: add)
saldos.insert(0, 20.0)      # insertar en una posición
print(saldos)

print(saldos[1], saldos[-1])  # acceso por índice
saldos[0] = 25.0              # modificar (Java: set)
saldos.remove(80.0)           # eliminar por valor
print(saldos)

print(len(saldos), sum(saldos), max(saldos), min(saldos))
print(sorted(saldos))         # nueva lista ordenada
print(250.5 in saldos)        # Java: contains
````

````python
estudiantes = ["Henry", "Jonnathan", "Ander", "Frixon", "Santiago"]
print(estudiantes)
estudiantes.append("Kevin")
print(estudiantes)
estudiantes.insert(1,"Alex")
print(estudiantes)
estudiantes.remove("Henry")
print(estudiantes)
print(estudiantes[5])
print("Numero de elementos: ",len(estudiantes))
print(estudiantes[::-1])
print(sorted(estudiantes))

````

El *slicing* también funciona con listas:

````python
meses = ["ene", "feb", "mar", "abr", "may", "jun"]
print(meses[:3])     # primer trimestre
print(meses[3:])     # segundo trimestre
print(meses[::-1])   # invertida
````

### 6.3 Diccionarios: `dict`
Un diccionario guarda pares **clave → valor**, como un `HashMap<String, Object>` de Java.

````python
socio = {
    "cedula": "0102030405",
    "nombre": "Ana Pérez",
    "saldo": 1520.75,
    "activo": True,
}

print(socio["nombre"])              # Java: map.get("nombre")
socio["saldo"] += 100               # modificar
socio["telefono"] = "0991234567"    # agregar una clave nueva (Java: put)
print(socio.get("correo", "sin correo"))  # valor por defecto si no existe

print(socio.keys())
print("cedula" in socio)            # Java: containsKey
````

````python
estudiante = {
    "nombre": "Jose Sigcha",
    "cedula": "0107332355",
    "edad":21,
    "direccion": "Tarqui",
    "notas": [9,7.8,6,7.9,8,6],
    "correo": "jose.sigcha.est@tecazuy.edu.ec"
}
print(estudiante["nombre"])
print(estudiante.keys())
print(estudiante["notas"][3])
print(estudiante.get("direccion"))
````

### 6.4 Conjuntos: `set`
Un conjunto **no admite repetidos**, como el `HashSet` de Java. Es útil para eliminar duplicados.

````python
tipos_cuenta = ["ahorros", "corriente", "ahorros", "plazo fijo", "ahorros"]
unicos = set(tipos_cuenta)
print(unicos, len(unicos))

cuentas_ana = {"ahorros", "plazo fijo"}
cuentas_luis = {"ahorros", "corriente"}
print(cuentas_ana & cuentas_luis)   # intersección: lo que tienen en común
print(cuentas_ana | cuentas_luis)   # unión
````

## 7. Mutabilidad y referencias (10 min)

| Inmutables (no cambian) | Mutables (sí cambian) |
|---|---|
| `int`, `float`, `bool`, `str`, `tuple` | `list`, `dict`, `set` |

Igual que con los objetos en Java, **una variable guarda una referencia**. Si dos variables apuntan a la misma lista, un cambio hecho con una se ve con la otra:

````python
cuentas_a = ["AH-001", "AH-002"]
cuentas_b = cuentas_a          # NO es una copia: es la misma lista
cuentas_b.append("AH-003")
print(cuentas_a)               # ¡también cambió!

copia = cuentas_a.copy()       # copia independiente
copia.append("AH-004")
print(cuentas_a, copia)
````

`==` compara el **contenido** (como `equals` en Java) e `is` compara si es **el mismo objeto** (como `==` con objetos en Java):

````python
x = [1, 2, 3]
y = [1, 2, 3]
print(x == y)   # True: mismo contenido
print(x is y)   # False: son dos listas distintas
````

## 8. Caso práctico: registro de un socio (10 min)

Vamos a representar a un socio de la cooperativa usando **todos los tipos vistos**:

````python
socio = {
    "cedula": "0102030405",                 # str: puede empezar con 0
    "nombre": "  ana lucía pérez ",          # str con espacios y minúsculas
    "edad": 34,                              # int
    "activo": True,                          # bool
    "garante": None,                         # None: todavía no tiene
    "cuentas": [                             # list de dict
        {"numero": "AH-001", "tipo": "ahorros", "saldo": 1520.75},
        {"numero": "PF-010", "tipo": "plazo fijo", "saldo": 5000.0},
        {"numero": "AH-002", "tipo": "ahorros", "saldo": 80.5},
    ],
}

#Actividad 1. Procesamiento de información personal
#Obtener el nombre del socio eliminando los espacios innecesarios.
sinespacios = socio["nombre"].strip()
print("Nombre sin espacios: ", sinespacios)

#Convertir el nombre a formato de título.
nombretitulo = sinespacios.title()
print("Nombre en formato titulo: ", nombretitulo)

#Mostrar la cédula y la edad.
print("Cedula: " + socio["cedula"])
print("Edad: " + str(socio["edad"]))

#Comprobar si el socio está activo.
actividad = socio["activo"]
print("El socio esta activo?: ", actividad)
````

````python
#Actividad 2. Procesamiento de cuentas

#Obtener una lista con los saldos de las cuentas utilizando comprensión de listas.
saldos = [socio["cuentas"][0]["saldo"], socio["cuentas"][1]["saldo"], socio["cuentas"][2]["saldo"]]
print("Lista de saldos: ", saldos)

#Obtener una lista con los números de cuenta.
numeros = [socio["cuentas"][0]["numero"], socio["cuentas"][1]["numero"], socio["cuentas"][2]["numero"]]
print("Lista de numeros de cuenta: ", numeros)

#Identificar los tipos de cuentas sin valores duplicados utilizando set.
unicos = set([socio["cuentas"][0]["tipo"], socio["cuentas"][1]["tipo"], socio["cuentas"][2]["tipo"]])
print(f"Tipos sin duplicados: {unicos}")

#Determinar cuántas cuentas tiene el socio utilizando len().
cuentas = len(socio["cuentas"])
print(f"Cantidad de cuentas: {cuentas}")
````

````python
#Actividad 3. Operaciones financieras

#Calcular el saldo total utilizando sum().
saldototal = sum([socio["cuentas"][0]["saldo"], socio["cuentas"][1]["saldo"], socio["cuentas"][2]["saldo"]])
print("Saldo total: ", saldototal)

#Identificar el saldo máximo utilizando max().
maximo = max([socio["cuentas"][0]["saldo"], socio["cuentas"][1]["saldo"], socio["cuentas"][2]["saldo"]])
print("Saldo maximo: ", maximo)

#Identificar el saldo mínimo utilizando min().
minimo = min([socio["cuentas"][0]["saldo"], socio["cuentas"][1]["saldo"], socio["cuentas"][2]["saldo"]])
print("Saldo minimo: ", minimo)

#Calcular el saldo promedio utilizando sum() y len().
saldos = [socio["cuentas"][0]["saldo"], socio["cuentas"][1]["saldo"], socio["cuentas"][2]["saldo"]]
promedio = sum(saldos) / len(saldos)
print("Saldo promedio: ", promedio)
````

````python
#Actividad 4. Validación de información

#Comprobar si existe un garante registrado.
garante = bool(socio["garante"])
print("Garante registrado?: ", garante)

#Mostrar un mensaje dependiendo del estado del socio.
estado = "Socio Activo" if socio["activo"] else "Socio Inactivo"
print("Estado del socio: ", estado)

#Verificar si el saldo total es mayor a $1.000.
saldototal = (socio["cuentas"][0]["saldo"] + socio["cuentas"][1]["saldo"] + socio["cuentas"][2]["saldo"])
calculo = saldototal > 1000
print("Es mayor a $1.000?: ", calculo)

#Determinar si existe al menos una cuenta de plazo fijo.
plazofijo = ("plazo fijo" in socio["cuentas"][0]["tipo"] or "plazo fijo" in socio["cuentas"][1]["tipo"] or "plazo fijo" in socio["cuentas"][2]["tipo"])
print("Existe unca cuenta de plazo fijo?: ", plazofijo)
````

| Entero / decimal | `int`, `long` / `double` | `int` (sin límite) / `float` |
| Sin valor | `null` | `None` |
| Texto | `String`, `char` | `str` (no hay `char`) |
| Dinero exacto | `BigDecimal` | `Decimal` |
| Subcadena | `substring(0, 4)` | `s[0:4]` |
| Lista dinámica | `ArrayList` | `list` → `[ ]` |
| Mapa | `HashMap` | `dict` → `{clave: valor}` |
| Conjunto | `HashSet` | `set` → `{ }` |
| Tipo de un objeto | `instanceof` | `isinstance()` |

## Soluciones de los ejercicios

````python
# Ejercicio 1
dato = "  0102030405 ; MARÍA josé torres ; 250.50  "
cedula, nombre, saldo = [p.strip() for p in dato.split(";")]
nombre = nombre.title()
saldo = float(saldo) + 49.50
print(nombre)
print(f"Saldo: ${saldo:.2f}")
print("Cédula válida:", cedula.isdigit() and len(cedula) == 10)
````

````python
# Ejercicio 2
depositos = [120.0, 45.5, 300.0, 45.5, 80.0]
depositos.append(200.0)
total = sum(depositos)
print(f"Total: ${total:.2f}")
print(f"Promedio: ${total / len(depositos):.2f}")
print("Mayor:", max(depositos), "| Menor:", min(depositos))
print("Montos distintos:", len(set(depositos)))
````

````python
# Ejercicio 3
tasas = {"ahorros": 0.02, "plazo fijo": 0.07, "corriente": 0.0}
print(f"Plazo fijo: {tasas['plazo fijo']:.1%}")
tasas["infantil"] = 0.03
print(tasas.get("vivienda", 0.0))
print(tasas)
````
